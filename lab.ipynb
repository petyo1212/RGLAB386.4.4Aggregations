```python
import numpy as np
import pandas as pd
```

```python
data = {

    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],

    'Value': [10, 15, 20, 25, 30, 35,40,50],

    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],

    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],

    'Discount':[1000,2300,1200,2500,2000,2000,500,600]

}
df=pd.DataFrame(data)
print(df)
```

```python
# Total sum column of value column using aggregate function
summ=df["Value"].sum()
total_sum=df['Value'].agg('sum')
print("Total sum: ", total_sum)
print("Sum: ", summ)
```

```python
# Cumulative sum
df['Value'].agg('cumsum')
```

```python
df[['Value']].cumsum()
```

```python
# Both agg and alt min and max and count value column
# min
m_s= df['Value'].min()
m_a=df[['Value']].agg('min')
print ("Function result: ", m_s)
print("Aggregate: ", m_a)
```

```python
# ax
mm_s= df['Value'].max()
mm_a=df[['Value']].agg('max')
print ("Function result: ", mm_s)
print("Aggregate: ", mm_a)
```

```python
# count
c_s= df['Value'].count()
c_a=df[['Value']].agg('count')
print ("Function result: ", c_s)
print("Aggregate: ", c_a)
```

```python
# Multiple aggregations
result= df[['Fee',"Discount"]].agg('sum')
result1= df[['Fee',"Discount"]].sum()
print(result)
print(result1)
```

```python
# Miltiple columns and aggregations
result3= df[['Fee',"Discount"]].agg(['sum','max','mean'])
print(result3)
```

```python
# Group by and aggregate function
result_group=df.groupby('Category')[["Value","Fee","Discount"]].agg(['sum','max'])
print(result_group)
```
